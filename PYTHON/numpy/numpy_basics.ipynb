# NumPy Basics

This notebook introduces NumPy arrays, operations, reshaping, and statistics.

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
print(arr)
print(arr.shape)
print(arr.dtype)
```

```python
matrix = np.array([[1, 2], [3, 4], [5, 6]])
print(matrix)
print(matrix.reshape(2, 3))
print(matrix.sum())
```

```python
data = np.array([10, 20, 30, 40, 50])
print('Mean:', np.mean(data))
print('Median:', np.median(data))
print('Standard Deviation:', np.std(data))
print('Min:', np.min(data), 'Max:', np.max(data))
```

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print(np.dot(A, B))
print(A + B)
print(A * B)
```
