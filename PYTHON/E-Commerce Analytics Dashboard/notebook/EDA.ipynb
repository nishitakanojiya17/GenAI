```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    y="Status",
    order=df["Status"].value_counts().index
)

plt.title("Order Status Distribution")
plt.xlabel("Number of Orders")
plt.ylabel("Status")

plt.tight_layout()
plt.show()
```

```python
average_order_value = df["Amount"].mean()

print("Average Order Value:", average_order_value)
```

```python
amount = df["Amount"].dropna().to_numpy()

print("Mean:", np.mean(amount))
print("Median:", np.median(amount))
print("Maximum:", np.max(amount))
print("Minimum:", np.min(amount))
print("Standard Deviation:", np.std(amount))
```

```python
category_sales = (
    df.groupby("Category")["Amount"]
    .sum()
    .sort_values(ascending=False)
)

category_sales
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=category_sales.values,
    y=category_sales.index
)

plt.title("Sales by Product Category")
plt.xlabel("Sales Amount")
plt.ylabel("Category")

plt.tight_layout()
plt.show()
```

Monthly Sales analysis 

```python
df["Month"] = df["Date"].dt.to_period("M")
```

```python
monthly_sales = (
    df.groupby("Month")["Amount"]
    .sum()
)
```

```python
monthly_sales
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales Amount")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

State-wise sales

```python
state_sales = (
    df.groupby("ship-state")["Amount"]
    .sum()
    .sort_values(ascending=False)
)
```

```python
state_sales.head(10)
```

```python
top_states = state_sales.head(10)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_states.values,
    y=top_states.index
)

plt.title("Top 10 States by Sales")
plt.xlabel("Sales Amount")
plt.ylabel("State")

plt.tight_layout()
plt.show()
```
